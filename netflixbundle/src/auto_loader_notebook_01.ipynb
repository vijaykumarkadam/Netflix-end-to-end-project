```python
checkpoint_location = "abfss://bronze@devnetflixdatalake.dfs.core.windows.net/checkpoint"
```

```python
df = spark.readStream.format("cloudFiles")\
        .option("cloudFiles.format", "csv")\
        .option("cloudFiles.schemaLocation", checkpoint_location)\
        .load("abfss://raw@devnetflixdatalake.dfs.core.windows.net/source")
```

```python
display(df)
```

```python
df.writeStream.format("parquet")\
        .option("checkpointLocation", checkpoint_location)\
        .trigger(once=True)\
        .start("abfss://bronze@devnetflixdatalake.dfs.core.windows.net/netflix_titles")
```
